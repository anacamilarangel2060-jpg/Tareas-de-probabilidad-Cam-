**Ejercicio 2**

En una contraseña de 3 dígitos, del 0 al 9, calcula la probabilidad de que ninguno de los dígitos se repita.

Una contraseña de 3 dígitos (usando números del o al 9) tiene 10x10x10 = 1000 combinaciones totales posibles. Para que ningún número se repita, el primer dígito tiene 10 opciones disponibles, el segundo tiene 9 opciones (excluyendo el primer número), el tercer dígito tiene 8 opciones (excluyendo lo dos anteriores) por lo cual los casos favorables son 10x9x8=720. La probabilidad es:

P=$\frac{720}{1000}$
= $\frac{18}{25}$ = 0.72

```python
from itertools import product
# S: Todas las combinaciones posibles de 3 dígitos (del 0 al 9)
S = list(product(range(10), repeat=3))
# A: Casos favorables que no tienen dígitos repetidos
A = [x for x in S if len(set(x)) == 3]

print("-----Contraseña de 3 dígitos")
print("Resultados posibles:", len(S))
print("Casos favorables:", len(A))
print("Probabilidad:", len(A) / len(S))
```

**Ejercicio 3**

Tienes 8 canciones distintas y las pones en un orden aleatorio, con todos los órdenes igualmente probables. Calcula la probabilidad de que tu favorita quede primera y la que menos te gusta quede última.

El número total de formas de ordenar 8 canciones distintas es 8! (40,320). Si fijamos la canción favorita en la primera posición y la que menos nos gusta en la última, quedan 6 canciones para organizar en las posiciones intermedias. Estas 6 canciones se pueden ordenar de 6! formas (720). La probabilidad es:
P= $\frac{6!}{8!}$= $\frac{6!}{(8)(7)(6!)}$ = 156

```python
from itertools import permutations
# C: Todas las permutaciones posibles de las 8 canciones
# Representadas del 1 al 8
playlist_total = list(permutations(range(1, 9)))
# D: Casos favorables donde la canción favorita (1) está al inicio y la menos favorita (8) al final
playlist_exitos = [playlist for playlist  in playlist_total if playlist[0] == 1 and playlist[-1] == 8]
print("-----Playlist de canciones-----")
print("Resultados posibles:", len(playlist_total))
print("Casos favorables:", len(playlist_exitos))
print("Probabilidad:", len(playlist_exitos) / len(playlist_total))
```

Se lanzan tres dados equilibrados. ¿Qué es más probable, que la suma sea 9 o que la suma sea 10? ¿Por cuántos resultados le gana una suma a la otra?
Como cada dado tiene 6 caras, el número total de resultados posibles es:  (6)(6)(6)=216

Resultados cuya suma es 9 Hay 25 resultados posibles:
(1,2,6): 6 permutaciones = {(1, 2, 6), (1, 6, 2), (2, 1, 6), (2, 6, 1), (6, 1, 2), (6, 2, 1)}
(1,3,5): 6 permutaciones = {(1, 3, 5), (1, 5, 3), (3, 1, 5), (3, 5, 1), (5, 1, 3), (5, 3, 1)}
(1,4,4): 3 permutaciones = {(1, 4, 4), (4, 1, 4), (4, 4, 1)}
(2,2,5): 3 permutaciones = {(2, 2, 5), (2, 5, 2), (5, 2, 2)}
(2,3,4): 6 permutaciones = {(2, 3, 4), (2, 4, 3),(3, 2, 4), (3, 4, 2),(4, 2, 3), (4, 3, 2)}
(3,3,3): 1 permutación = {(3, 3, 3)}
Total:  (6+6+3+3+6+1=25)

Resultados cuya suma es 10 Hay 27 resultados posibles:
(1,3,6): 6 permutaciones = {(1, 3, 6), (1, 6, 3), (3, 1, 6), (3, 6, 1), (6, 1, 3), (6, 3, 1)}

(1,4,5): 6 permutaciones = {(1, 4, 5), (1, 5, 4), (4, 1, 5), (4, 5, 1), (5, 1, 4), (5, 4, 1)}

(2,2,6): 3 permutaciones = {(2, 2, 6), (2, 6, 2), (6, 2, 2)}

(2,3,5): 6 permutaciones = {(2, 3, 5), (2, 5, 3), (3, 2, 5), (3, 5, 2), (5, 2, 3), (5, 3, 2)}

(2,4,4): 3 permutaciones = {(2, 4, 4), (4, 2, 4), (4, 4, 2)}

```python
from itertools import product
# S: Lanzamiento de 3 dados de 6 caras
espacio_datos = list(product(range(1, 7), repeat=3))
# Tenemos que filtrar las combinaciones para las sumas 9 y 10
favorable_nueve = [x for x  in espacio_datos if sum(x)== 9]
favorable_diez = [x for x  in espacio_datos if sum(x) == 10]

print("----Lanzamiento de 3 dados-----")
print("Resultados posibles:", len(espacio_datos))
print("\nCasos favorables para 9:", len(favorable_nueve))
print("Combinaciones suma 9:", favorable_nueve)

print("\nCasos favorables para 10:", len(favorable_diez))
print("Combinaciones suma 10:", favorable_diez)
print("\n¿Por cuántos resultados le gana la suma 10? ")
direfencia = len(favorable_diez) - len(favorable_nueve)
print(direfencia)
```

**Ejercicio 5**

Ana, Bruno, Carla y Diego hacen un amigo secreto: cada quien saca un regalo al azar y todos los repartos son igualmente probables. Calcula la probabilidad de que nadie saque su propio regalo.



```python
from itertools import permutations
# Representamos a las 4 personas y regalos por sus índices: 0, 1, 2, 3
amigos = [0, 1, 2, 3]
# C: Todas las permutaciones posibles de entrega de regalos
intercambio_total = list(permutations(amigos))

# A: Casos favorables donde ningún amigo (i) recibe su propio regalo (p[i])
sin_autoregalo = [p for p in intercambio_total if all(p[i] != i for i in range(len(amigos)))]

print("-----Intercambio de Regalos-----")
print("Resultados posibles:", len(intercambio_total))
print("Casos favorables (nadie con su regalo):", len(sin_autoregalo))
print("Combinaciones favorables:", sin_autoregalo)
prob_amigo_secreto = len(sin_autoregalo) / len(intercambio_total)
print(f"Probabilidad final: {len(sin_autoregalo)}/{len(intercambio_total)} {prob_amigo_secreto:.4f}")
```
