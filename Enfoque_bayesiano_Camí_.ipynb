# Ejercicio 1

Se lanza un dado equilibrado. Calcula la probabilidad de obtener un 2, dado que salió un número par.

El evento B (que salga par) reduce el espacio muestral a {2, 4, 6}. De esos tres resultados, uno corresponde al evento A (obtener 2). Entonces $P(2 | par) = \frac{1}{3}$

```python
espacio = {1, 2, 3, 4, 5, 6}
B = {x for x in espacio if x % 2 ==0}
A = {x for x in espacio if x ==2}
interseccion = A & B
print(f"(P 2 | par) {len(interseccion) / len(B)} ")
```

Ejercicio 2
Se lanzan tres dados. Calcula la probabilidad de que la suma sea 9, dado que las tres caras son distintas.

El número de combinaciones con 3 caras distintas es 6x5x4=120. Las combinaciones sin importar el orden que suman 9 con números distintos son (1,2,5), (1,3,5), y (2,3,4). Estas combinaciones tienen 3!=6 permutaciones, por lo que en total hay 3x6=18



```python
from itertools import product

S = list(product(range(1, 7), repeat=3))
distintos = [t for t in S if len(set(t)) == 3]
suma_9 = [t for t in distintos if sum(t) == 9]

print("Resultados favorables / Casos condicionados:", len(suma_9), "/", len(distintos))
print("P(suma 9 | caras distintas) =", len(suma_9) / len(distintos))
```

# Ejercicio 3
De 100 días:

40 son nublados

De esos, 30 llueve

¿Cuál es la probabilidad de lluvia dado que está nublado?

Nos están dando directamente el total de casos que cumplen la condición (40 nublados) y cuántos de ellos también cumplen el evento deseado (30 con lluvia).

$P(\text{lluvia} \mid \text{nublado}) = \frac{30}{40} = \frac{3}{4} = 0.75$

```python
nublados = 40
llueve_nublado = 30
p_lluvia_dado_nublado = llueve_nublado / nublados
print(f"P(lluvia | nublado) = {p_lluvia_dado_nublado}")
```

# Ejercicio 1
El 70% de los clientes ven un anuncio y, de ellos, compra el 30%. El 30% no ve el anuncio y, de ellos, compra el 10%. Calcula la probabilidad total de que un cliente compre.

$P(\text{Compra}) = P(\text{Compra} \mid \text{Ve}) \cdot P(\text{Ve}) + P(\text{Compra} \mid \text{NoVe}) \cdot P(\text{NoVe})$

$P(\text{Compra}) = P(\text 0.30) \cdot P(\text 0.70) + P(\text 0.10) \cdot P(\text 0.30) = 0.21 + 0.03 = 0.24 $

La probabilidad es de 24 % de que un cliente compre.

```python
# Eventos: V = ve anuncio, NV = NO SE VE ANUNCIO
p_V = 0.70
p_NV = 0.30
# Probabilidades condicionales
p_C_V = 0.30
p_C_NV = 0.10
# Probabilidad total
p_C = (p_C_V * p_V) + (p_C_NV * p_NV)
print(f"Probabilidad total de compra:{p_C:.2f}")
```

# Ejercicio 2
Se sabe que:

El 40% de los días son nublados

De esos, el 75% llueve

El 60% de los días no son nublados

De esos, el 10% llueve

¿Cuál es la probabilidad total de que llueva?

$P(\text{Llueve}) = P(\text{Llueve} \mid \text{Nublado}) \cdot P(\text{Nublado}) + P(\text{Llueve} \mid \text{NoNublado}) \cdot P(\text{NoNublado})$

$P(\text{Llueve}) = P(\text 0.75) \cdot P(\text 0.40) + P(\text 0.10) \cdot P(\text 0.60) = 0.30 + 0.06 = 0.36$

Probabilidad: 36 % de probabilidad de que llueva

```python
# Estados de clima: N=Nublado, NN = No nublado
p_N = 0.40
p_NN = 0.60
# Probabilidades condicionales
p_L_N = 0.75
p_L_NN = 0.10
# Aplicando Teorema de proabilidad total: P(L) = P(L|N)P(N) + P(L|NN)P(NN)
p_L = (p_L_N * p_N) + (p_L_NN * p_NN)

print(f"Probabilidad total de lluvia: {p_L:.2f}")
```

# Ejercicio 3
Hay dos cajas de audífonos. La primera tiene 2 que sirven y 6 que fallan. La segunda tiene 5 que sirven y 3 que fallan. Se elige una caja al azar y después un audífono. ¿Cuál es la probabilidad de que sirva?

Como se elige al azar, cada caja tiene probabilidad de $\frac{1}{2}$ (0.5) de ser seleccionada. La caja 1 tiene probabilidad de que sirva de $\frac{2}{8}$ (0.25). La caja 2 tiene $\frac{5}{8}$ (0.625).

Fórmula: $P(\text{Sirve}) = (\frac{2}{8}) \cdot 0.5 + (\frac{5}{8}) \cdot 0.5 = 0.125 + 0.3125 = 0.4375$

La probabilidad de que el audífono sirva es de 43.75 %

```python
# Probabilidad de elegir cualquiera de las dos cajas al azar
p_eleccion_caja  = 0.5

# Probabilidad de extraer un audífono que sirva por cada caja
p_sirve_caja1 = 2 / 8
p_sirve_caja2 = 5 / 8

# Aplicando el Teorema de Probabilidad total
prob_total_audifono = (p_sirve_caja1 * p_eleccion_caja) + (p_sirve_caja2 * p_eleccion_caja)

print(f"Probabilidad de que el audífono sirva: {prob_total_audifono:.4f}")
```

# Teorema de Bayes

# Ejercicio 1
El 20% de los correos son spam. La palabra «oferta» aparece en el 80% de los correos spam y en el 10% de los correos deseados. Si un correo tiene la palabra «oferta», ¿cuál es la probabilidad de que sea spam?


$
P(S\mid O)=
\frac{P(O\mid S)\,P(S)}
{P(O\mid S)P(S)+P(O\mid D)P(D)}
$

Donde:
- (P(S)=0.20): probabilidad inicial de que sea spam.
- (P(O | S)=0.80): probabilidad de que un correo spam contenga «oferta».
- (P(D)=0.80): probabilidad de que sea deseado.
- (P(O | D)=0.10\): probabilidad de que un correo deseado contenga «oferta».

Sustituimos:
$P(S\mid O)=
\frac{(0.80)(0.20)}
{(0.80)(0.20)+(0.10)(0.80)}
$

Resolvemos las multiplicaciones:
$P(S\mid O)=\frac{0.16}{0.16+0.08}$

$ P(S\mid O)=\frac{0.16}{0.24}=0.6667 $

Probabilidad de que sea spam dado que contiene la palabra oferta = 66.67 %

```python
# Probabilidades: S = Spam, D = Deseado (No spam)
p_S = 0.20
p_D = 0.80
# Probabilidades condicionales: 0 = Contiene la palabra "oferta"
p_O_S = 0.80
p_O_D = 0.10
# Teorema de Bayes: P(Spam | Oferta)
p_S_O = (p_O_S * p_S) / (p_O_S * p_S + p_O_D * p_D)
print(f"P(Spam |  Oferta) = {p_S_O:.4f}({p_S_O * 100:.2f}%)")
```

# Ejercicio 2
En una tienda online, el 40% entra por Chrome, el 35% por Safari y el 25% por otros. La tasa de conversión (compra) es del 3% en Chrome, 4% en Safari y 2% en otros.

Si se registró una venta, ¿qué navegador es más probable que haya usado el cliente?

Sacamos la probabilidad total de Venta (V):
$P(V) = (0.03 \cdot 0.40) + (0.04 \cdot 0.35) + (0.02 \cdot 0.25) = 0.012 + 0.014 + 0.005 = 0.031$.

Calculamos Bayes para cada navegador (el navegador por la tasa de conversión entre la probabilidad total de venta):

Chrome: $\frac{0.012}{0.031} \approx 0.387$ (38.7%)

Safari: $\frac{0.014}{0.031} \approx 0.451$ (45.1%)

Otros: $\frac{0.005}{0.031} \approx 0.161$ 16.1%

El valor más alto corresponde a Safari, por lo tanto, es el navegador más probable en el que se hizo la venta.

```python
# Probabilidades a priorizar de cada navegador
p_Chrome = 0.40
p_Safari = 0.35
p_Otros  = 0.25
# Probabilidades de venta condicionales (tasas de conversión)
p_venta_Chrome = 0.03
p_venta_Safari = 0.04
p_venta_Otros = 0.02
# Probabilidad total de venta
p_venta = (p_venta_Chrome * p_Chrome) + (p_venta_Safari * p_Safari) + (p_venta_Otros * p_Otros)
# Aplicando Teorema de Bayes
p_Chrome_venta = (p_venta_Chrome * p_Chrome) / p_venta
p_Safari_venta = (p_venta_Safari * p_Safari) / p_venta
p_Otros_venta = (p_venta_Otros * p_Otros) / p_venta
# Imprimir resultados
print(f"Probabilidad de venta por Chrome: {p_Chrome_venta:.3f}")
print(f"Probabilidad de venta por Safari: {p_Safari_venta:.3f}")
print(f"Probabilidad de venta por Otros: {p_Otros_venta:.3f}")
# Conclución del navegador más probable
print("\nEl navegador más probable en el que se hizo la venta es: Safari")
```
