# Ejercicio 1

Simula 10 000 lanzamientos de dos dados y estima la probabilidad de que la suma sea exactamente 5. Compara esa frecuencia con la probabilidad clásica que ya calculaste.

Para simular 10,000 lanzamientos y buscar la suma de 5, generamos dos números aleatorios del 1 al 6 en cada iteración.
$$P(A) = \frac{4}{36} = \frac{1}{9} \approx 0.11111$$

```python
import random
import matplotlib.pyplot as plt

def tirar_par_dados():
  # Suma de dos valores aleatorios del 1 al 6
  return random.randint(1, 6) + random.randint(1, 6)
# Configuración de los parámetros de  simulación
total_tiradas = 10000
aciertos_suma_cinco = 0
historial_frecuencias = []

# Bucle para simular el experimento de forma progresiva
for paso in range(1, total_tiradas + 1):
  resultado_tiro = tirar_par_dados()
# Validamos si la suma obtenida es exactamente 5
  if resultado_tiro == 5:
    aciertos_suma_cinco += 1
  # Calculamos la frecuencia relativa actual y lo guardó en la lista
  frecuencia_relativa = aciertos_suma_cinco /paso
  historial_frecuencias.append(aciertos_suma_cinco / paso)

print("---Simulación frecuentista: Suma de 5 ---")
print("Total de tiradas realizadas:",total_tiradas)
print("Probabilidad empírica obtenida:", frecuencia_relativa)
```

```python
# Generación de la gráfica
plt.figure(figsize=(10,5))
plt.plot(historial_frecuencias, label="Frecuencia empírics (Suma 5)", color="purple")
plt.axhline(1/9, color="blue", linestyle="--", label="Probabilidad teórica = 1/9")

plt.xlabel("Número de tiradas")
plt.ylabel("Frecuencia relativa")
plt.title("Convergencia frecuentista: Simulación de Suma 5 en Dos Dados")
plt.legend()
plt.show()
```

# Ejercicio 2

Simula 10 000 PIN de 4 dígitos, del 0 al 9, y estima la probabilidad de que ninguno se repita. Compara el resultado con la fracción de la parte clásica.

Usa generar_pin del ejemplo de los tres dígitos iguales, ahora con 4 dígitos.

$$P(A) = \frac{10 \times 9 \times 8 \times 7}{10^4} = \frac{5,040}{10,000} = \frac{63}{125} = 0.504$$

```python
import random
import matplotlib.pyplot as plt
# Creó una función para generar un PIN aleatorio de 4 dígitos
def crear_codigo_pin():
  return [random.randint(0, 9) for _ in range(4)]
n = 10000
conteo_sin_repeticion = 0
historial_frecuencias = []

for paso in range(1, n + 1):
  codigo_pin = crear_codigo_pin()
  if len(set(codigo_pin)) == 4:
    conteo_sin_repeticion += 1
  frecuencia_relativa = conteo_sin_repeticion / paso
  historial_frecuencias.append(frecuencia_relativa)

print("---Simulación frecuentista: Sin repeticiones ---")
print("Total de tiradas realizadas:", n)
print("Probabilidad empírica obtenida:", frecuencia_relativa)

```

```python
# Ahora creó la gráfica
plt.figure(figsize=(10,5))
plt.plot(historial_frecuencias, label="Frecuencia empírica (Sin repetición)", color="darkcyan")
plt.axhline(5040/10000, color="deeppink", linestyle="--", label="Probabilidad teórica = 5040/10000")

plt.xlabel("Número de tiradas")
plt.ylabel("Frecuencia relativa")
plt.title("Convergencia frecuentista: Simulación de PIN sin repetición")
plt.legend()
plt.show()
```

# Ejercicio 3

Simula 20 000 lanzamientos de tres dados. Estima la frecuencia de la suma 9 y la de la suma 10. ¿Se nota, en la simulación, cuál de las dos sale más?

*Un ensayo son tres llamadas a lanzar_dado. Lleva dos conteos. La ventaja de una suma sobre la otra es pequeña: si en una corrida casi empatan, sube el número de lanzamientos y vuelve a correr.

Ejecutaremos 20,000 simulaciones para comparar ambas sumas. En la probabilidad clásica, la suma de 10 gana por un pequeño margen (27 contra 25 casos de 216).

```python
import random
import matplotlib.pyplot as plt

def lanzar_tres_dados():
  return random.randint(1, 6) + random.randint(1, 6) + random.randint(1, 6)

n = 20000
conteo_nueve = 0
conteo_diez = 0
historial_frecuencias = []
frecuencias_nueve = []
frecuencias_diez = []
for paso in range(1, n + 1):
  suma_actual = lanzar_tres_dados()

  if suma_actual == 9:
    conteo_nueve += 1
  elif suma_actual == 10:
    conteo_diez += 1
  frecuencias_nueve.append(conteo_nueve / paso)
  frecuencias_diez.append(conteo_diez / paso)

print("---Simulación frecuentista: Suma de 9 y 10 ---")
print("Total de tiradas realizadas:", n)
print("Probabilidad empírica de la suma 9:", conteo_nueve / n)
print("Probabilidad empírica de la suma 10:", conteo_diez / n)
```

```python
# Creamos la gráfica correctamente alineada
plt.figure(figsize=(10, 5))
plt.plot(frecuencias_diez, label="Suma 10 (Teórica = 27/216)", color="darkorange")
plt.plot(frecuencias_nueve, label="Suma 9 (Teórica = 25/216)", color="royalblue")

plt.xlabel("Número de tiradas")
plt.ylabel("Frecuencia relativa")
plt.title("Convergencia frecuentista: Simulación de Suma 9 y 10 en Tres Dados")
plt.legend()
plt.show()
```

# Ejercicio 4

Basado en el Ejemplo 3, simula una urna con 3 bolas rojas y 2 azules. Realiza 2,000 extracciones (con reemplazo) y calcula la probabilidad de obtener una azul.

```python
import random

n_extracciones = 2000
conteo_azules = 0
# Urna con 3 rojas y dos azules
urna = ["rojo", "rojo", "rojo", "azul", "azul"]

for _ in range(n_extracciones):
  # Extrae una bola al azar de la urna
  bola_extraida = random.choice(urna)
  if bola_extraida == "azul":
    conteo_azules += 1
probabilidad_empirica = conteo_azules / n_extracciones

print("----Simulación de urna----")
print("Total de extracciones:", n_extracciones)
print("Probabilidad empírica de obtener una bola azul:", probabilidad_empirica)
print("Probabilidad Teórica exacta: 2/5 (0.4)")
```

# Ejercicio 5

Una serie son 20 volados. Estima, con 10 000 series, la probabilidad de que aparezcan 5 águilas seguidas en algún momento de la serie.

Un ensayo no es un volado: son 20 llamadas a lanzar_moneda. Recorre esa serie y cuenta el ensayo si la racha de águilas llega a 5. La estimación es el número de series favorables entre 10 000. Aquí no hay una fracción simple de la parte clásica contra la cual dibujar la línea.

```python
import random
n_series = 10000
series_favorables = 0

for _ in range(n_series):
  # Generar una serie de 20 volados ('A' = águila, 'S' = sol)
   serie = [random.choice(['A', 'S']) for _ in range(20)]

   racha_actual = 0
   tiene_racha_de_cinco = False
   for volado in serie:
     if volado == 'A':
       racha_actual += 1
       if racha_actual == 5:
         tiene_racha_de_cinco = True
         break
     else:
        racha_actual = 0
   if tiene_racha_de_cinco:
     series_favorables += 1
probabilidad_empirica = series_favorables / n_series

print("----Simulación de volados----")
print("Total de series:", n_series)
print(" Series con racha de cinco águilas:", series_favorables)
print("Probabilidad estimada:", probabilidad_empirica)
```

# Ejercicio 6 — Bonus

Los clientes llegan a una tienda de forma aleatoria (entre 1 y 5 minutos). El cajero tarda entre 2 y 4 minutos en atender. Simula la llegada de 500 clientes y calcula la probabilidad de que un cliente tenga que esperar más de 10 minutos en la fila.

```python
import random

n_clientes = 500
tiempo_llegada_actual = 0
tiempo_cajero_libre = 0
clientes_espera_mayor_10 = 0

for i in range(n_clientes):
    # Tiempo transcurrido desde el cliente anterior (1 a 5 min)
    interllegada = random.uniform(1, 5)
    tiempo_llegada_actual += interllegada

    # Tiempo de servicio de este cliente (2 a 4 min)
    tiempo_servicio = random.uniform(2, 4)

    # Si llega cuando el cajero está ocupado, espera en fila
    if tiempo_llegada_actual < tiempo_cajero_libre:
        espera = tiempo_cajero_libre - tiempo_llegada_actual
    else:
        espera = 0

    if espera > 10:
        clientes_espera_mayor_10 += 1

    # El cajero estará libre después de atender a este cliente
    tiempo_inicio_atencion = max(tiempo_llegada_actual, tiempo_cajero_libre)
    tiempo_cajero_libre = tiempo_inicio_atencion + tiempo_servicio

probabilidad_espera_10 = clientes_espera_mayor_10 / n_clientes

print("--- Simulación de cola en tienda (Bonus) ---")
print("Total de clientes simulados:", n_clientes)
print("Clientes que esperaron más de 10 minutos:", clientes_espera_mayor_10)
print("Probabilidad estimada:", probabilidad_espera_10)
```
